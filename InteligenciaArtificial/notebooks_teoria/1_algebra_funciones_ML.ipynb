## 🔢 Variables numéricas y operaciones básicas

En esta primera parte aprenderás cómo declarar variables numéricas en Python y realizar operaciones simples como suma, resta, multiplicación y división.

Estas operaciones serán la base para entender cómo funcionan los modelos de machine learning, donde cada número representa un dato o una predicción.

```python
# Declaramos dos variables numéricas
x = 5
y = 2

# Realizamos operaciones básicas
suma = x + y
resta = x - y
multiplicacion = x * y
division = x / y

# Imprimimos los resultados
print("Suma:", suma)
print("Resta:", resta)
print("Multiplicación:", multiplicacion)
print("División:", division)
```

## 🧮 Introducción a vectores como listas

Un vector es simplemente una lista ordenada de números. En Machine Learning, los vectores representan características de los datos, como la edad, ingresos, temperatura, etc.

En Python, podemos usar listas para trabajar con vectores.

```python
# Lista como vector
vector = [2, 4, 6, 8]

# Acceso a elementos
print("Primer elemento:", vector[0])

# Sumar todos los elementos del vector
suma = sum(vector)
print("Suma de elementos:", suma)
```

## 📐 Introducción a matrices con NumPy

Las matrices son arreglos bidimensionales de números. Representan conjuntos de datos en forma de tabla.

Utilizaremos la biblioteca `numpy`, que es muy popular para cálculos numéricos en Python.

```python
import numpy as np

# Crear una matriz 2x2
matriz = np.array([[1, 2], [3, 4]])
print("Matriz:\n", matriz)

# Crear otra matriz 2x2
matriz2 = np.array([[5, 6], [7, 8]])

# Suma de matrices
suma = matriz + matriz2
print("Suma de matrices:\n", suma)

# Producto punto de matrices (dot product)
producto = np.dot(matriz, matriz2)
print("Producto punto:\n", producto)
```

## 📈 Funciones lineales y gráficas

En Machine Learning, muchas relaciones entre variables se modelan como funciones lineales del tipo:

$$
y = mx + b
$$

Donde:
- `m` es la pendiente
- `b` es el intercepto o punto de corte


```python
import numpy as np
import matplotlib.pyplot as plt

# Definir una función lineal
def funcion_lineal(x):
    return 2 * x + 3

# Crear valores para x en el rango de -10 a 10
x = np.linspace(-10, 10, 100)
y = funcion_lineal(x)

# Graficar la función
plt.plot(x, y)
plt.title("Función lineal: y = 2x + 3")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

## 🔗 Correlación visual entre variables

En Machine Learning es común visualizar si existe una relación entre variables.

Por ejemplo, podemos generar datos sintéticos donde `y` depende de `x` con algo de "ruido" aleatorio.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generar datos aleatorios correlacionados
np.random.seed(0)  # para reproducibilidad
x = np.random.rand(50) * 10  # valores entre 0 y 10
y = 2 * x + np.random.randn(50) * 2  # y depende de x con algo de ruido

# Graficar dispersión
plt.scatter(x, y)
plt.title("Relación entre variables")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

## 🧠 Producto punto aplicado a regresión lineal

En Machine Learning, los datos son vectores y el modelo aprende un vector de pesos.  
La predicción se hace con el **producto punto**:

$$
\hat{y} = \vec{w}^T \vec{x}
$$


```python
import numpy as np

# Vector de entrada (características)
x = np.array([1, 2])  # ejemplo: [temperatura, festivo]

# Vector de pesos del modelo
w = np.array([0.5, 1.5])

# Producto punto para obtener predicción
pred = np.dot(x, w)
print("Predicción (producto punto):", pred)
```
