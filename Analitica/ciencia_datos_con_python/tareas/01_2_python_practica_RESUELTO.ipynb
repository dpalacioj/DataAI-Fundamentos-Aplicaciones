## 6. ¿Cuál es la diferencia entre una Tupla y una Lista?

**Objetivo:** Entender las diferencias fundamentales entre estructuras de datos.

### Respuesta:

**Principales diferencias entre Tuplas y Listas:**

| Característica | Lista | Tupla |
|----------------|-------|-------|
| **Mutabilidad** | ✅ Mutable (se puede modificar) | ❌ Inmutable (no se puede modificar) |
| **Sintaxis** | Corchetes `[]` | Paréntesis `()` |
| **Velocidad** | Más lenta | Más rápida |
| **Uso de memoria** | Mayor | Menor |
| **Uso típico** | Datos que cambian | Datos que no cambian |
| **Puede ser llave de diccionario** | No | Sí |

```python
# Demostración práctica

# LISTA - Mutable
mi_lista = [1, 2, 3]
print(f"Lista original: {mi_lista}")
mi_lista[0] = 100  # ✅ Se puede modificar
mi_lista.append(4)  # ✅ Se puede agregar elementos
print(f"Lista modificada: {mi_lista}")

print("\n" + "="*50 + "\n")

# TUPLA - Inmutable
mi_tupla = (1, 2, 3)
print(f"Tupla original: {mi_tupla}")

# Intentar modificar una tupla genera error
try:
    mi_tupla[0] = 100  # ❌ No se puede modificar
except TypeError as e:
    print(f"❌ Error al intentar modificar tupla: {e}")

# Las tuplas pueden ser llaves de diccionarios
diccionario = {(1, 2): "coordenada", (3, 4): "otra coordenada"}
print(f"\n✅ Tuplas como llaves de diccionario: {diccionario}")
```

## 8. Cuente las ocurrencias de la letra `x` en la siguiente lista

**Objetivo:** Practicar el método `count()` y diferencias entre mayúsculas/minúsculas.

```python
# Lista dada
lista = ["a", "b", "c", "x", "x", "X", "x"]
print(f"Lista: {lista}")

# Solución 1: Contar solo 'x' minúscula
conteo_x_minuscula = lista.count("x")
print(f"\n✅ Ocurrencias de 'x' (minúscula): {conteo_x_minuscula}")

# Solución 2: Contar 'X' mayúscula
conteo_X_mayuscula = lista.count("X")
print(f"Ocurrencias de 'X' (mayúscula): {conteo_X_mayuscula}")

# Solución 3: Contar tanto 'x' como 'X' (ignorando caso)
conteo_total = sum(1 for letra in lista if letra.lower() == "x")
print(f"\n📊 Ocurrencias totales (ignorando mayúsculas/minúsculas): {conteo_total}")

# Análisis detallado
print("\n📋 Análisis detallado:")
for i, letra in enumerate(lista):
    if letra.lower() == "x":
        print(f"  Posición [{i}]: '{letra}'")
```

## 11. Función `domainGet`. Cree una función llamada domainGet que extraiga el dominio de un correo electrónico

**Objetivo:** Practicar funciones y manipulación de strings.

```python
# Solución 1: Usando split()
def domainGet(email):
    """
    Extrae el dominio de un correo electrónico.
    
    Args:
        email (str): Dirección de correo electrónico
    
    Returns:
        str: Dominio del correo
    """
    return email.split('@')[1]

# Solución 2: Usando slicing con find()
def domainGet_v2(email):
    """Versión alternativa usando find()"""
    indice_arroba = email.find('@')
    return email[indice_arroba + 1:]

# Solución 3: Con validación
def domainGet_v3(email):
    """Versión con validación de formato"""
    if '@' not in email:
        return "Error: No es un email válido"
    return email.split('@')[1]

# Pruebas
emails_prueba = [
    "juan_perez@correo.com",
    "maria@gmail.com",
    "contacto@empresa.co",
    "usuario@subdominio.dominio.com"
]

print("✅ Resultados:")
for email in emails_prueba:
    print(f"  {email} → {domainGet(email)}")
```

## 12. Función `buscar_palabra_promedio`

**Objetivo:** Practicar búsqueda de strings ignorando mayúsculas/minúsculas.

```python
# Solución 1: Usando 'in' con lower()
def palabra_promedio(texto):
    """
    Busca si la palabra 'promedio' está en el texto (ignorando mayúsculas).
    
    Args:
        texto (str): Texto donde buscar
    
    Returns:
        bool: True si contiene 'promedio', False si no
    """
    return 'promedio' in texto.lower()

# Solución 2: Usando find()
def palabra_promedio_v2(texto):
    """Versión usando find()"""
    return texto.lower().find('promedio') != -1

# Pruebas
textos_prueba = [
    "El Promedio de la clase es de 4.3",
    "El PROMEDIO es importante",
    "Este texto no contiene la palabra",
    "El promedio, mediana y moda son medidas estadísticas"
]

print("✅ Resultados:")
for texto in textos_prueba:
    resultado = palabra_promedio(texto)
    emoji = "✅" if resultado else "❌"
    print(f"  {emoji} {resultado}: '{texto[:40]}...'")
```

## 13. Función contar_python

**Objetivo:** Contar ocurrencias de una palabra específica.

```python
# Solución 1: Usando count() con lower()
def contar_python(texto):
    """
    Cuenta cuántas veces aparece la palabra 'python' (ignorando mayúsculas).
    
    Args:
        texto (str): Texto donde contar
    
    Returns:
        int: Número de ocurrencias de 'python'
    """
    return texto.lower().count('python')

# Solución 2: Usando split() para contar palabras completas
def contar_python_v2(texto):
    """
    Versión que cuenta solo palabras completas 'python'
    (no cuenta si 'python' es parte de otra palabra)
    """
    palabras = texto.lower().split()
    return sum(1 for palabra in palabras if 'python' in palabra)

# Solución 3: Usando expresiones regulares (más precisa)
import re

def contar_python_v3(texto):
    """
    Versión con regex para contar solo palabras completas
    """
    patron = r'\bpython\b'
    return len(re.findall(patron, texto, re.IGNORECASE))

# Pruebas
textos_prueba = [
    "Python es un lenguaje de programacion, Python es divertido",
    "PYTHON python PyThOn",
    "Me gusta programar en Python. Python es versátil.",
    "pythonista usa python para todo"
]

print("✅ Resultados:")
for texto in textos_prueba:
    conteo = contar_python(texto)
    print(f"  Ocurrencias: {conteo} en '{texto[:50]}...'")
```

## 14. 💣 Detección de anomalías

**Objetivo:** Implementar lógica condicional compleja para detectar anomalías estadísticas.

```python
def deteccion_anomalia(valor, valor_media, desviacion_estandar, distribucion_normal):
    """
    Detecta anomalías en un conjunto de datos basándose en la regla 3-sigma.
    
    La regla 3-sigma establece que en una distribución normal:
    - ~68% de los datos están dentro de ±1σ de la media
    - ~95% de los datos están dentro de ±2σ de la media
    - ~99.7% de los datos están dentro de ±3σ de la media
    
    Args:
        valor (float): El valor a evaluar
        valor_media (float): La media del conjunto de datos
        desviacion_estandar (float): La desviación estándar
        distribucion_normal (bool): Si la distribución es normal
    
    Returns:
        str: "anormal", "normal" o "indeterminado"
    """
    
    # Si no es distribución normal, no podemos aplicar la regla 3-sigma
    if not distribucion_normal:
        return "indeterminado"
    
    # Calcular los límites superior e inferior (3 sigmas)
    limite_superior = valor_media + 3 * desviacion_estandar
    limite_inferior = valor_media - 3 * desviacion_estandar
    
    # Verificar si el valor está fuera de los límites
    if valor > limite_superior:
        return "anormal"
    elif valor < limite_inferior:
        return "anormal"  # Nota: El enunciado dice "normal" pero debería ser "anormal"
    else:
        return "normal"

# Versión corregida según el enunciado exacto
def deteccion_anomalia_original(valor, valor_media, desviacion_estandar, distribucion_normal):
    """
    Versión según el enunciado original (con la lógica especificada).
    """
    if not distribucion_normal:
        return "indeterminado"
    
    limite_superior = valor_media + 3 * desviacion_estandar
    limite_inferior = valor_media - 3 * desviacion_estandar
    
    # Según el enunciado:
    if valor > limite_superior:
        return "anormal"
    elif valor < limite_inferior:
        return "normal"  # Esto parece un error en el enunciado
    else:
        return "indeterminado"

# Pruebas
print("✅ Pruebas con la función corregida:")
print("="*50)

# Test 1
resultado1 = deteccion_anomalia_original(
    valor=15.6, 
    valor_media=9.5, 
    desviacion_estandar=8.7, 
    distribucion_normal=True
)
print(f"Test 1: {resultado1} (se espera: 'normal')")
print(f"  Análisis: 15.6 está en [{9.5-3*8.7:.1f}, {9.5+3*8.7:.1f}]")

# Test 2
resultado2 = deteccion_anomalia_original(
    valor=45.4, 
    valor_media=9.5, 
    desviacion_estandar=8.7, 
    distribucion_normal=True
)
print(f"\nTest 2: {resultado2} (se espera: 'anormal')")
print(f"  Análisis: 45.4 > {9.5+3*8.7:.1f} (límite superior)")

# Test 3
resultado3 = deteccion_anomalia_original(
    valor=15.6, 
    valor_media=9.5, 
    desviacion_estandar=8.7, 
    distribucion_normal=False
)
print(f"\nTest 3: {resultado3} (se espera: 'indeterminado')")
print(f"  Análisis: distribución_normal = False")

# Visualización de la regla 3-sigma
print("\n📊 Regla 3-Sigma:")
print("  68% de datos: μ ± 1σ")
print("  95% de datos: μ ± 2σ")
print("  99.7% de datos: μ ± 3σ")
print("  Valores fuera de μ ± 3σ son considerados anomalías")
```
