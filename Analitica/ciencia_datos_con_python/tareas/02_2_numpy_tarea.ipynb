## Ejercicio 1: Distribución de manzanas
Imagina que tienes 10 manzanas y quieres distribuirlas equitativamente entre 4 amigos. Usa NumPy para calcular cuántas manzanas recibe cada amigo y cuántas sobran.

```python
import numpy as np

# Número total de manzanas
manzanas = 10
# Número de amigos
amigos = 4

# Distribuir manzanas
manzanas_por_amigo = np.floor(manzanas / amigos)
sobrantes = manzanas % amigos

print(f"Cada amigo recibe {manzanas_por_amigo} manzanas y sobran {sobrantes} manzanas.")
```

## Ejercicio 2: Precios de frutas en el mercado
Tienes una lista de precios de diferentes frutas en el mercado por kilogramo: [3.5, 2.8, 4.2, 5.0, 3.0]. Usa NumPy para encontrar el precio medio, el precio más alto y el precio más bajo.

```python
precios = np.array([3.5, 2.8, 4.2, 5.0, 3.0])

precio_medio = np.mean(precios)
precio_maximo = np.max(precios)
precio_minimo = np.min(precios)

print(f"El precio medio es {precio_medio}, el precio más alto es {precio_maximo} y el precio más bajo es {precio_minimo}.")
```

## Ejercicio 3: Calificación de estudiantes
Imagina que tienes las calificaciones de 5 estudiantes en una prueba: [85, 90, 78, 92, 88]. Usa NumPy para calcular la calificación promedio, la calificación más alta y la calificación más baja.

```python
calificaciones = np.array([85, 90, 78, 92, 88])

promedio = np.mean(calificaciones)
calificacion_maxima = np.max(calificaciones)
calificacion_minima = np.min(calificaciones)

print(f"La calificación promedio es {promedio}, la calificación más alta es {calificacion_maxima} y la calificación más baja es {calificacion_minima}.")
```

## Ejercicio 4: Temperaturas diarias
Tienes un array de temperaturas diarias registradas durante una semana: [22, 21, 23, 20, 19, 24, 25]. Usa NumPy para encontrar la temperatura promedio de la semana y la diferencia entre la temperatura más alta y la más baja.

```python
temperaturas = np.array([22, 21, 23, 20, 19, 24, 25])

temp_promedio = np.mean(temperaturas)
diferencia_temp = np.max(temperaturas) - np.min(temperaturas)

print(f"La temperatura promedio de la semana es {temp_promedio}°C y la diferencia entre la temperatura más alta y la más baja es {diferencia_temp}°C.")
```

## Ejercicio 5: Ahorros mensuales
Estás ahorrando dinero cada mes y tienes la siguiente cantidad ahorrada por mes durante el año: [200, 250, 300, 350, 400, 450, 500, 550, 600, 650, 700, 750]. Usa NumPy para calcular el total ahorrado en el año y el promedio mensual de ahorro.

```python
ahorros = np.array([200, 250, 300, 350, 400, 450, 500, 550, 600, 650, 700, 750])

total_ahorrado = np.sum(ahorros)
promedio_ahorro_mensual = np.mean(ahorros)

print(f"El total ahorrado en el año es ${total_ahorrado} y el promedio mensual de ahorro es ${promedio_ahorro_mensual}.")
```

## Ejercicio 6: Distribución de notas de un examen
Imagina que tienes las siguientes notas de un examen: [55, 75, 85, 90, 100, 60, 70, 80, 95, 85]. Usa NumPy para calcular la desviación estándar y la varianza de las notas.

```python
notas = np.array([55, 75, 85, 90, 100, 60, 70, 80, 95, 85])

desviacion_estandar = np.std(notas)
varianza = np.var(notas)

print(f"La desviación estándar de las notas es {desviacion_estandar} y la varianza es {varianza}.")
```
