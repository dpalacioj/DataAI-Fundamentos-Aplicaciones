# Estadística Inferencial - Conceptos Básicos

Este notebook introduce conceptos fundamentales de estadística inferencial.

## 1. Población

### Definición
Conjunto completo de elementos que poseen características de interés para un estudio.

### Características
- **Población finita**: Número limitado de elementos
- **Población infinita**: Número ilimitado de elementos

### Muestra
Subconjunto representativo de la población utilizado para hacer inferencias.

## 2. Media

### Media Poblacional (μ)
Promedio de todos los valores en la población.

$$\mu = \frac{\sum_{i=1}^{N} x_i}{N}$$

### Media Muestral (x̄)
Promedio de los valores en una muestra.

$$\bar{x} = \frac{\sum_{i=1}^{n} x_i}{n}$$

### Propiedades
- Medida de tendencia central
- Sensible a valores extremos
- Base para muchos procedimientos estadísticos

## 3. Tipos de Distribuciones

### Distribución Normal
- **Forma**: Campana simétrica
- **Parámetros**: μ (media), σ (desviación estándar)
- **Aplicación**: Fenómenos naturales, errores de medición

### Distribución Binomial
- **Contexto**: Experimentos con dos resultados posibles
- **Parámetros**: n (ensayos), p (probabilidad de éxito)
- **Aplicación**: Control de calidad, encuestas

### Distribución de Poisson
- **Contexto**: Eventos raros en intervalos de tiempo/espacio
- **Parámetro**: λ (tasa promedio de ocurrencia)
- **Aplicación**: Llegada de clientes, defectos por unidad

### Distribución Uniforme
- **Característica**: Todos los valores tienen igual probabilidad
- **Parámetros**: a (mínimo), b (máximo)
- **Aplicación**: Números aleatorios, tiempos de espera

## 4. Aplicaciones Prácticas con SciPy.stats

A continuación, exploramos el uso de la librería `scipy.stats` con datos sintéticos para ilustrar los conceptos anteriores.

````python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
from scipy import stats

# Configuración general
np.random.seed(42)
````

### 4.1 Trabajando con Distribución Normal

````python
# Crear datos sintéticos de una población normal
# Simulamos alturas de estudiantes universitarios (en cm)
poblacion_alturas = stats.norm.rvs(loc=170, scale=10, size=10000, random_state=42)

# Tomar una muestra
muestra_alturas = np.random.choice(poblacion_alturas, size=100, replace=False)

# Calcular estadísticos
print("POBLACIÓN vs MUESTRA")
print(f"Media poblacional: {poblacion_alturas.mean():.2f} cm")
print(f"Media muestral: {muestra_alturas.mean():.2f} cm")
print(f"Desviación estándar poblacional: {poblacion_alturas.std():.2f}")
print(f"Desviación estándar muestral: {muestra_alturas.std():.2f}")

# Visualización con Plotly
fig = make_subplots(rows=1, cols=2, 
                    subplot_titles=('Población (N=10,000)', 'Muestra (n=100)'))

# Histograma de población
fig.add_trace(
    go.Histogram(x=poblacion_alturas, nbinsx=50, name='Población',
                 histnorm='probability density', 
                 marker_color='blue', opacity=0.7),
    row=1, col=1
)

# Línea vertical para media poblacional
fig.add_vline(x=poblacion_alturas.mean(), line_dash="dash", line_color="red",
              annotation_text=f"μ = {poblacion_alturas.mean():.1f}",
              row=1, col=1)

# Histograma de muestra
fig.add_trace(
    go.Histogram(x=muestra_alturas, nbinsx=20, name='Muestra',
                 histnorm='probability density',
                 marker_color='green', opacity=0.7),
    row=1, col=2
)

# Línea vertical para media muestral
fig.add_vline(x=muestra_alturas.mean(), line_dash="dash", line_color="red",
              annotation_text=f"x̄ = {muestra_alturas.mean():.1f}",
              row=1, col=2)

fig.update_xaxes(title_text="Altura (cm)", row=1, col=1)
fig.update_xaxes(title_text="Altura (cm)", row=1, col=2)
fig.update_yaxes(title_text="Densidad", row=1, col=1)
fig.update_yaxes(title_text="Densidad", row=1, col=2)

fig.update_layout(height=400, showlegend=False,
                  title_text="Comparación Población vs Muestra")
fig.show()
````

### 4.2 Prueba de Normalidad

````python
# Prueba de Shapiro-Wilk para normalidad
# H0: Los datos siguen una distribución normal
# H1: Los datos NO siguen una distribución normal

# Datos normales
datos_normales = stats.norm.rvs(size=100, random_state=42)
shapiro_stat_normal, p_value_normal = stats.shapiro(datos_normales)

# Datos no normales (exponencial)
datos_exponencial = stats.expon.rvs(size=100, random_state=42)
shapiro_stat_exp, p_value_exp = stats.shapiro(datos_exponencial)

print("PRUEBA DE SHAPIRO-WILK")
print("-" * 40)
print("Datos Normales:")
print(f"  Estadístico: {shapiro_stat_normal:.4f}")
print(f"  p-valor: {p_value_normal:.4f}")
print(f"  Conclusión: {'No rechazar H0 - Datos normales' if p_value_normal > 0.05 else 'Rechazar H0 - Datos no normales'}")
print("\nDatos Exponenciales:")
print(f"  Estadístico: {shapiro_stat_exp:.4f}")
print(f"  p-valor: {p_value_exp:.4f}")
print(f"  Conclusión: {'No rechazar H0 - Datos normales' if p_value_exp > 0.05 else 'Rechazar H0 - Datos no normales'}")
````

### 4.4 Intervalo de Confianza para la Media

#### ¿Qué es un Intervalo de Confianza?

Un **intervalo de confianza** es un rango de valores que, con cierta probabilidad (ej. 95%), contiene el verdadero valor del parámetro poblacional que estamos estimando.

#### ¿Qué muestra el gráfico?

El siguiente gráfico muestra **20 experimentos diferentes** donde:
- Cada línea horizontal = una muestra diferente de 30 estudiantes
- Cada punto = la media de esa muestra específica  
- Cada línea = el rango donde "probablemente" está la media poblacional real
- La línea azul vertical = la media poblacional verdadera (μ=75)

#### ¿Qué significa que "no contiene μ"?

Cuando una línea es **roja**, significa que ese intervalo específico **NO incluye** la media poblacional real (75). Esto pasa porque:

1. **Es probabilístico**: Un intervalo del 95% significa que esperamos que ~5% fallen
2. **Mala suerte del muestreo**: Esa muestra particular no fue representativa  
3. **Es normal**: En 20 muestras, esperamos que ~1 falle (5% de 20 = 1)

#### ¿Cuándo se usa?

- **Estimación de parámetros**: Cuando no conocemos la población completa
- **Toma de decisiones**: Para evaluar rangos plausibles de valores
- **Comunicación de incertidumbre**: Para mostrar qué tan seguros estamos

#### ¿Por qué es importante en Machine Learning?

**1. Evaluación de Modelos**
```python
# En lugar de decir "precisión = 85%"
# Decimos "precisión = 85% ± 3%" (intervalo: 82%-88%)
```

**2. Comparación de Algoritmos**
- ¿Es realmente mejor el Modelo A vs Modelo B?
- Si los intervalos se superponen, la diferencia puede no ser significativa

**3. Confianza en Predicciones**
- Una predicción de "precio = $100,000" es menos útil que
- "precio = $100,000 ± $15,000 (IC 95%)"

**4. Validation Sets**
- Los resultados de validación cruzada incluyen intervalos de confianza
- Ayudan a detectar overfitting o modelos inestables

**En resumen**: Los intervalos de confianza nos dicen "qué tan seguros estamos" de nuestras estimaciones, algo crucial para tomar decisiones basadas en datos.

````python
# Simulación: Calificaciones de un examen
# Población real con media = 75 y desviación = 12
poblacion_calificaciones = stats.norm.rvs(loc=75, scale=12, size=5000, random_state=42)

# Tomar varias muestras y calcular intervalos de confianza
n_muestras = 20
tamaño_muestra = 30
nivel_confianza = 0.95

resultados = []

for i in range(n_muestras):
    muestra = np.random.choice(poblacion_calificaciones, size=tamaño_muestra, replace=False)
    media_muestra = muestra.mean()
    error_estandar = stats.sem(muestra)
    
    # Intervalo de confianza usando t-distribution
    intervalo = stats.t.interval(nivel_confianza, 
                                 df=tamaño_muestra-1, 
                                 loc=media_muestra, 
                                 scale=error_estandar)
    
    resultados.append({
        'muestra': i+1,
        'media': media_muestra,
        'IC_inferior': intervalo[0],
        'IC_superior': intervalo[1],
        'contiene_media_real': intervalo[0] <= 75 <= intervalo[1]
    })

df_resultados = pd.DataFrame(resultados)

# Visualización de intervalos de confianza con Plotly
fig = go.Figure()

# Añadir intervalos de confianza
for idx, row in df_resultados.iterrows():
    color = 'green' if row['contiene_media_real'] else 'red'
    
    # Línea del intervalo
    fig.add_trace(go.Scatter(
        x=[row['IC_inferior'], row['IC_superior']],
        y=[row['muestra'], row['muestra']],
        mode='lines',
        line=dict(color=color, width=3),
        showlegend=False
    ))
    
    # Punto de la media
    fig.add_trace(go.Scatter(
        x=[row['media']],
        y=[row['muestra']],
        mode='markers',
        marker=dict(color=color, size=8),
        showlegend=False
    ))

# Línea vertical para media poblacional
fig.add_vline(x=75, line_dash="dash", line_color="blue", line_width=3,
              annotation_text="Media poblacional real (μ=75)")

# Añadir leyenda manual
fig.add_trace(go.Scatter(x=[None], y=[None], mode='markers',
                        marker=dict(color='green', size=10),
                        name='Contiene μ'))
fig.add_trace(go.Scatter(x=[None], y=[None], mode='markers',
                        marker=dict(color='red', size=10),
                        name='No contiene μ'))

fig.update_layout(
    xaxis_title="Calificación",
    yaxis_title="Número de muestra",
    title=f"Intervalos de Confianza al {nivel_confianza*100}% para {n_muestras} muestras",
    height=500,
    showlegend=True
)

fig.show()

# Resumen
cobertura = df_resultados['contiene_media_real'].mean() * 100
print(f"RESUMEN DE INTERVALOS DE CONFIANZA")
print(f"-" * 40)
print(f"Nivel de confianza teórico: {nivel_confianza*100}%")
print(f"Cobertura observada: {cobertura:.1f}%")
print(f"Intervalos que contienen μ: {df_resultados['contiene_media_real'].sum()}/{n_muestras}")
````

### 4.5 Prueba t de Student

````python
# Comparación de dos grupos: Efecto de un tratamiento
# Grupo control vs Grupo tratamiento

# Generar datos sintéticos
np.random.seed(42)
grupo_control = stats.norm.rvs(loc=100, scale=15, size=50)
grupo_tratamiento = stats.norm.rvs(loc=108, scale=15, size=50)  # Efecto real de 8 puntos

# Prueba t para muestras independientes
t_statistic, p_value = stats.ttest_ind(grupo_control, grupo_tratamiento)

# Estadísticos descriptivos
control_mean = grupo_control.mean()
control_std = grupo_control.std()
tratamiento_mean = grupo_tratamiento.mean()
tratamiento_std = grupo_tratamiento.std()
diferencia = tratamiento_mean - control_mean

print("PRUEBA T DE STUDENT - MUESTRAS INDEPENDIENTES")
print("=" * 50)
print(f"Grupo Control:")
print(f"  Media: {control_mean:.2f}")
print(f"  Desv. Est.: {control_std:.2f}")
print(f"  n: {len(grupo_control)}")
print(f"\nGrupo Tratamiento:")
print(f"  Media: {tratamiento_mean:.2f}")
print(f"  Desv. Est.: {tratamiento_std:.2f}")
print(f"  n: {len(grupo_tratamiento)}")
print(f"\nDiferencia de medias: {diferencia:.2f}")
print(f"\nResultados de la prueba:")
print(f"  Estadístico t: {t_statistic:.4f}")
print(f"  p-valor: {p_value:.4f}")
print(f"  Conclusión: {'Diferencia significativa (p < 0.05)' if p_value < 0.05 else 'No hay diferencia significativa (p ≥ 0.05)'}")

# Visualización con Plotly
fig = make_subplots(rows=1, cols=2,
                    subplot_titles=('Comparación de Grupos', 'Distribución de los Grupos'))

# Boxplot comparativo
df_grupos = pd.DataFrame({
    'Control': grupo_control,
    'Tratamiento': grupo_tratamiento
})
df_melted = df_grupos.melt(var_name='Grupo', value_name='Valor')

fig.add_trace(
    go.Box(y=df_melted[df_melted['Grupo']=='Control']['Valor'],
           name='Control', marker_color='lightblue'),
    row=1, col=1
)
fig.add_trace(
    go.Box(y=df_melted[df_melted['Grupo']=='Tratamiento']['Valor'],
           name='Tratamiento', marker_color='lightcoral'),
    row=1, col=1
)

# Histogramas superpuestos
fig.add_trace(
    go.Histogram(x=grupo_control, name='Control', 
                 histnorm='probability density',
                 marker_color='blue', opacity=0.5, nbinsx=15),
    row=1, col=2
)
fig.add_trace(
    go.Histogram(x=grupo_tratamiento, name='Tratamiento',
                 histnorm='probability density', 
                 marker_color='red', opacity=0.5, nbinsx=15),
    row=1, col=2
)

# Líneas verticales para medias
fig.add_vline(x=control_mean, line_dash="dash", line_color="blue",
              line_width=2, row=1, col=2)
fig.add_vline(x=tratamiento_mean, line_dash="dash", line_color="red",
              line_width=2, row=1, col=2)

fig.update_yaxes(title_text="Valor", row=1, col=1)
fig.update_xaxes(title_text="Valor", row=1, col=2)
fig.update_yaxes(title_text="Densidad", row=1, col=2)

fig.update_layout(height=500, title_text="Análisis de Prueba t de Student")
fig.show()
````

### 4.6 Cálculo de Percentiles y Cuartiles

````python
# Datos sintéticos: Salarios en una empresa (en miles)
np.random.seed(42)
salarios = np.concatenate([
    stats.norm.rvs(loc=50, scale=10, size=70),   # Empleados junior
    stats.norm.rvs(loc=80, scale=15, size=25),   # Empleados senior
    stats.norm.rvs(loc=120, scale=20, size=5)    # Ejecutivos
])

# Calcular percentiles usando scipy.stats
percentiles = [10, 25, 50, 75, 90, 95, 99]
valores_percentiles = [stats.scoreatpercentile(salarios, p) for p in percentiles]

# Crear DataFrame para mejor visualización
df_percentiles = pd.DataFrame({
    'Percentil': percentiles,
    'Valor (miles)': valores_percentiles
})

print("ANÁLISIS DE PERCENTILES - SALARIOS")
print("=" * 50)
print(df_percentiles.to_string(index=False))

# Estadísticos adicionales
print(f"\nEstadísticos descriptivos:")
print(f"  Media: ${salarios.mean():.2f}k")
print(f"  Mediana (P50): ${np.median(salarios):.2f}k")
print(f"  Rango intercuartílico (IQR): ${valores_percentiles[3] - valores_percentiles[1]:.2f}k")

# Visualización con Plotly
fig = make_subplots(rows=1, cols=2,
                    subplot_titles=('Distribución de Salarios con Cuartiles', 'Boxplot de Salarios'))

# Histograma con percentiles marcados
fig.add_trace(
    go.Histogram(x=salarios, nbinsx=30, name='Salarios',
                 marker_color='steelblue', opacity=0.7),
    row=1, col=1
)

# Líneas verticales para cuartiles
cuartiles = [25, 50, 75]
colores = ['orange', 'red', 'green']
for i, (p, v, color) in enumerate(zip(cuartiles, [valores_percentiles[1], valores_percentiles[2], valores_percentiles[3]], colores)):
    fig.add_vline(x=v, line_dash="dash", line_color=color,
                  annotation_text=f"P{p}: ${v:.0f}k",
                  row=1, col=1)

# Boxplot
fig.add_trace(
    go.Box(y=salarios, name='Salarios', marker_color='lightgreen',
           boxmean=True),
    row=1, col=2
)

fig.update_xaxes(title_text="Salario (miles)", row=1, col=1)
fig.update_yaxes(title_text="Frecuencia", row=1, col=1)
fig.update_yaxes(title_text="Salario (miles)", row=1, col=2)

fig.update_layout(height=500, showlegend=False,
                  title_text="Análisis de Percentiles y Cuartiles")
fig.show()
````

### 4.7 Pruebas de Hipótesis en Machine Learning

#### Problema: ¿Influye la experiencia laboral en la aprobación de préstamos?

Vamos a investigar si la **experiencia laboral** (en años) es una variable importante para predecir si un préstamo bancario será **aprobado o rechazado**.

````python
# Generar datos sintéticos de solicitudes de préstamo
np.random.seed(42)
n_solicitudes = 1000

# Crear datos sintéticos con una relación real pero con ruido
experiencia_años = np.random.exponential(scale=5, size=n_solicitudes)  # Distribución sesgada
experiencia_años = np.clip(experiencia_años, 0, 40)  # Límite realista

# Probabilidad de aprobación aumenta con experiencia (pero con ruido)
# Simulamos que tener más experiencia SÍ influye en la aprobación
prob_aprobacion = 0.3 + 0.02 * experiencia_años + np.random.normal(0, 0.1, n_solicitudes)
prob_aprobacion = np.clip(prob_aprobacion, 0, 1)

# Generar aprobaciones basadas en probabilidades
aprobado = np.random.binomial(1, prob_aprobacion)

# Crear DataFrame
df_prestamos = pd.DataFrame({
    'experiencia_años': experiencia_años,
    'aprobado': aprobado,
    'estado': ['Aprobado' if x == 1 else 'Rechazado' for x in aprobado]
})

print("DATASET DE PRÉSTAMOS BANCARIOS")
print("=" * 50)
print(f"Total de solicitudes: {len(df_prestamos)}")
print(f"Préstamos aprobados: {df_prestamos['aprobado'].sum()}")
print(f"Tasa de aprobación: {df_prestamos['aprobado'].mean():.1%}")
print(f"\nEstadísticos de experiencia:")
print(f"  Media: {df_prestamos['experiencia_años'].mean():.1f} años")
print(f"  Mediana: {df_prestamos['experiencia_años'].median():.1f} años")
print(f"  Desv. Est.: {df_prestamos['experiencia_años'].std():.1f} años")

# Mostrar primeras filas
print(f"\nPrimeras 10 solicitudes:")
print(df_prestamos.head(10).to_string(index=False))
````

#### Formulación de Hipótesis

**Pregunta de investigación**: ¿Existe diferencia significativa en la experiencia laboral entre personas que obtienen préstamos aprobados vs rechazados?

**H₀ (Hipótesis nula)**: No hay diferencia en la experiencia promedio entre grupos  
μ_aprobados = μ_rechazados

**H₁ (Hipótesis alternativa)**: Sí hay diferencia en la experiencia promedio  
μ_aprobados ≠ μ_rechazados

**Nivel de significancia**: α = 0.05

````python
# Análisis exploratorio por grupos
aprobados = df_prestamos[df_prestamos['aprobado'] == 1]['experiencia_años']
rechazados = df_prestamos[df_prestamos['aprobado'] == 0]['experiencia_años']

# Estadísticos descriptivos
print("ANÁLISIS POR GRUPOS")
print("=" * 50)
print(f"Préstamos APROBADOS (n={len(aprobados)}):")
print(f"  Media: {aprobados.mean():.2f} años")
print(f"  Mediana: {aprobados.median():.2f} años")
print(f"  Desv. Est.: {aprobados.std():.2f} años")

print(f"\nPréstamos RECHAZADOS (n={len(rechazados)}):")
print(f"  Media: {rechazados.mean():.2f} años")
print(f"  Mediana: {rechazados.median():.2f} años")
print(f"  Desv. Est.: {rechazados.std():.2f} años")

print(f"\nDiferencia de medias: {aprobados.mean() - rechazados.mean():.2f} años")

# Prueba t de Student para muestras independientes
t_statistic, p_value = stats.ttest_ind(aprobados, rechazados)

print(f"\nPRUEBA T DE STUDENT")
print(f"-" * 30)
print(f"Estadístico t: {t_statistic:.4f}")
print(f"p-valor: {p_value:.6f}")
print(f"Conclusión: {'RECHAZAR H₀ - La experiencia SÍ influye' if p_value < 0.05 else 'NO RECHAZAR H₀ - La experiencia NO influye significativamente'}")

# Tamaño del efecto (Cohen's d)
pooled_std = np.sqrt(((len(aprobados)-1)*aprobados.var() + (len(rechazados)-1)*rechazados.var()) / (len(aprobados)+len(rechazados)-2))
cohens_d = (aprobados.mean() - rechazados.mean()) / pooled_std
print(f"Tamaño del efecto (Cohen's d): {cohens_d:.3f}")
print(f"Interpretación: {'Efecto pequeño' if abs(cohens_d) < 0.5 else 'Efecto mediano' if abs(cohens_d) < 0.8 else 'Efecto grande'}")
````
