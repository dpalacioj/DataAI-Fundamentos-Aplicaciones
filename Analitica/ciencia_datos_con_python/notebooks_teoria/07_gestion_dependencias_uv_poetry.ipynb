# Gestión de Dependencias: UV y Poetry

Este notebook explica cómo configurar herramientas modernas para la gestión de dependencias de Python.

## 1. ¿Qué es UV?

**UV** es un gestor de paquetes y proyectos Python extremadamente rápido, escrito en Rust por Astral.

### Características principales:
- **Velocidad**: 10-100x más rápido que pip
- **Gestión de versiones**: Instala y gestiona versiones de Python automáticamente
- **Resolución de dependencias**: Algoritmo avanzado para resolver conflictos
- **Compatibilidad**: Drop-in replacement para pip y pip-tools

## 2. Instalación de UV

### Windows
```powershell
# Opción 1: PowerShell (recomendado)
powershell -c "irm https://astral.sh/uv/install.ps1 | iex"

# Opción 2: Scoop
scoop install uv

# Opción 3: Chocolatey
choco install uv

# Opción 4: pip
pip install uv
```

### macOS
```bash
# Opción 1: Script oficial (recomendado)
curl -LsSf https://astral.sh/uv/install.sh | sh

# Opción 2: Homebrew
brew install uv

# Opción 3: pip
pip install uv
```

## 3. ¿Qué es Poetry?

**Poetry** es una herramienta para gestión de dependencias y empaquetado en Python.

### Características principales:
- **Archivo único**: pyproject.toml para toda la configuración
- **Lock file**: poetry.lock garantiza builds reproducibles
- **Gestión de entornos virtuales**: Automática
- **Publicación**: Facilita la publicación en PyPI

## 4. Instalación de Poetry

### Windows
```powershell
# PowerShell (recomendado)
(Invoke-WebRequest -Uri https://install.python-poetry.org -UseBasicParsing).Content | py -

# pip (alternativa)
pip install poetry
```

### macOS
```bash
# Script oficial (recomendado)
curl -sSL https://install.python-poetry.org | python3 -

# Homebrew
brew install poetry

# pip (alternativa)
pip install poetry
```

## 5. Configuración Post-Instalación

### UV - Verificación
```bash
# Verificar instalación
uv --version

# Ver comandos disponibles
uv --help
```

### Poetry - Configuración
```bash
# Verificar instalación
poetry --version

# Configurar Poetry para crear entornos virtuales en el proyecto
poetry config virtualenvs.in-project true

# Ver configuración
poetry config --list
```

## 6. Comandos Básicos

### UV - Gestión de Proyectos
```bash
# Crear nuevo proyecto
uv init mi-proyecto

# Instalar dependencias
uv add pandas numpy scipy

# Instalar dependencias de desarrollo
uv add --dev pytest black flake8

# Ejecutar comandos en el entorno
uv run python script.py
uv run jupyter notebook

# Sincronizar entorno
uv sync
```

### Poetry - Gestión de Proyectos
```bash
# Crear nuevo proyecto
poetry new mi-proyecto

# Inicializar en proyecto existente
poetry init

# Instalar dependencias
poetry add pandas numpy scipy

# Instalar dependencias de desarrollo
poetry add --group dev pytest black flake8

# Instalar todas las dependencias
poetry install

# Ejecutar comandos
poetry run python script.py
poetry run jupyter notebook

# Activar shell
poetry shell
```

## 8. Recomendaciones por Caso de Uso

**Recomendado: UV o Poetry**
- Reproducibilidad garantizada

### Para Investigación Científica 🔬
**Recomendado: Miniconda**
- Paquetes optimizados para rendimiento

### Para DevOps/Producción 🚀
**Recomendado: UV**
- Velocidad en construcción de imágenes

## 7. Ventajas sobre Anaconda

### UV vs Anaconda
| Aspecto | UV | Anaconda |
|---------|----|-----------|
| **Velocidad** | ⚡ Extremadamente rápido | 🐌 Lento |
| **Tamaño** | 📦 ~50MB | 💾 ~3GB |
| **Gestión Python** | ✅ Automática | ⚠️ Manual |
| **Ecosistema** | 🐍 PyPI nativo | 🔄 conda-forge |
| **Resolución** | 🧠 Inteligente | ❓ Básica |

### Poetry vs Anaconda
| Aspecto | Poetry | Anaconda |
|---------|--------|-----------|
| **Configuración** | 📝 pyproject.toml | 📄 environment.yml |
| **Lock files** | 🔒 poetry.lock | ❌ No nativo |
| **Publicación** | 📤 PyPI integrado | ⚠️ Complejo |
| **Entornos** | 🎯 Por proyecto | 🌐 Global |
| **Reproducibilidad** | ✅ Garantizada | ⚠️ Variable |

### Ventajas principales

**Rendimiento**:
- UV es significativamente más rápido para instalación de paquetes
- Menor uso de memoria y espacio en disco
- Resolución de dependencias más eficiente

**Modernidad**:
- Siguen estándares PEP modernos (PEP 517, 621, 631)
- Mejor integración con herramientas de desarrollo
- Flujo de trabajo más simple y directo

**Simplicidad**:
- Un solo archivo de configuración (pyproject.toml)
- Comandos más intuitivos
- Menos configuración manual requerida

## 8. Cuándo usar cada herramienta

### Usar UV cuando:
- ✅ Necesitas velocidad máxima
- ✅ Trabajas principalmente con PyPI
- ✅ Quieres gestión automática de Python
- ✅ Proyectos de desarrollo web/aplicaciones

### Usar Poetry cuando:
- ✅ Necesitas publicar paquetes en PyPI
- ✅ Quieres máximo control sobre dependencias
- ✅ Trabajas en bibliotecas Python
- ✅ Requieres builds reproducibles

### Usar Anaconda cuando:
- ✅ Necesitas paquetes científicos con bibliotecas C/C++
- ✅ Trabajas con R, Julia o Scala además de Python
- ✅ Requieres GUI (Anaconda Navigator)
- ✅ Institución ya estandarizada en conda

## 9. Migración desde Anaconda

### Exportar dependencias de conda
```bash
# Desde ambiente conda activo
conda list --export > requirements.txt

# O usando conda env
conda env export > environment.yml
```

### Importar a UV
```bash
# Crear proyecto UV
uv init

# Instalar desde requirements.txt
uv add -r requirements.txt
```

### Importar a Poetry
```bash
# Inicializar Poetry
poetry init

# Agregar dependencias manualmente o
# Usar herramientas como conda-env-to-poetry
```

2. **Separar dependencias**: Producción vs desarrollo
3. **Versiones específicas**: Para reproducibilidad
4. **Documentar**: README con instrucciones de setup
