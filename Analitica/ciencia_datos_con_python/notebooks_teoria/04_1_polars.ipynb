[Polars](https://pola.rs/) es una biblioteca de manipulación de datos de alto rendimiento desarrollada por [Ritchie Vink](https://www.linkedin.com/in/ritchievink/). Inicialmente, Polars fue diseñado para ofrecer una alternativa más rápida y eficiente a pandas, utilizando técnicas modernas de procesamiento paralelo y optimización de memoria. Polars está escrito en Rust, lo que le permite manejar grandes volúmenes de datos de manera eficiente y segura, aprovechando al máximo los recursos del hardware disponible.

Los principales tipos de datos que pueden representarse con Polars son:

* Datos tabulares con columnas de tipo heterogéneo con etiquetas en columnas y filas.
* Series temporales.

Polars proporciona herramientas que permiten:

* Leer y escribir datos en diferentes formatos: CSV, JSON, Parquet y más.
* Seleccionar y filtrar tablas de datos de manera sencilla en función de posición, valor o etiquetas.
* Fusionar y unir datos.
* Transformar datos aplicando funciones tanto globalmente como por ventanas.
* Manipulación de series temporales.
* Hacer gráficas.

En Polars existen 2 tipos básicos de objetos:

* Series (listas, 1D)
* DataFrame (tablas, 2D)

Polars nos proporciona las estructuras de datos y funciones necesarias para el análisis de datos de una manera eficiente y rápida, aprovechando las ventajas del procesamiento paralelo y la optimización de memoria. Esto hace que Polars sea una excelente opción para proyectos que requieren manejar grandes volúmenes de datos y realizar análisis complejos de manera eficiente.

## Diferencias principales entre Polars y Pandas

| **Aspecto**                 | **Pandas**                                                                                                                                                                                   | **Polars**                                                                                                                                                                                                           |
|-----------------------------|----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|-----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| **Lenguaje subyacente**     | Construido originalmente sobre Python y NumPy (C en parte), pero la gestión de memoria puede verse limitada por la forma en la que Python maneja ciertos tipos de datos (por ejemplo, categóricos).                              | Escrito en Rust, lo que permite un control más eficiente de la memoria y un mejor aprovechamiento del hardware, similar a la velocidad de C o C++.                                                                    |
| **Rendimiento**             | Generalmente más lento cuando se trabaja con grandes volúmenes de datos o tipos de datos categóricos (pueden presentarse cuellos de botella en operaciones masivas).                                                              | Excelente rendimiento en operaciones intensivas gracias al paralelismo automático y la optimización interna, logrando procesar millones de filas de forma más veloz.                                                  |
| **Backend / Formato interno** | Pandas (hasta la versión 1.x) usaba principalmente estructuras basadas en NumPy. Con Pandas 2.0, adopta Arrow, pero usando la paquetería Pyarrow.                                                                                  | Utiliza Arrow como base, pero implementado directamente en Rust. Esto brinda gran interoperabilidad y estandarización de memoria, a la vez que mantiene un alto desempeño.                                            |
| **Evaluación**              | Evaluación inmediata (eager): cada operación se ejecuta al momento, lo que puede generar gastos computacionales repetidos (parciales) en la misma cadena de transformaciones.                                                     | Ofrece evaluación diferida (lazy), lo que acumula y optimiza las operaciones antes de ejecutarlas, reduciendo de manera significativa el tiempo total de procesamiento.                                               |
| **Uso de memoria**          | Puede ser más propenso a utilizar grandes cantidades de memoria en operaciones complejas y grandes volúmenes de datos.                                                                                                            | Maneja la memoria de forma más eficiente y optimizada, reduciendo la probabilidad de cuellos de botella y aprovechando mejor la RAM disponible.                                                                      |
| **Ecosistema y madurez**    | Amplio soporte y popularidad: muy integrado con librerías como scikit-learn, matplotlib, etc. Es la herramienta estándar para muchas tareas de análisis de datos en Python.                                                      | Cada vez más popular, en rápido desarrollo y con buena documentación. Ofrece integración creciente con otras librerías y se perfila como una alternativa robusta para el análisis de datos de alto rendimiento.         |

## Series

Una serie es el primer tipo de datos de Polars y es similar a una matriz NumPy. Lo que diferencia una matriz NumPy de una serie en Polars es que una serie puede tener etiquetas en los ejes, lo que significa que puede ser indexada por una etiqueta, en lugar de solo una ubicación numérica. Además, no necesita contener datos numéricos; puede contener cualquier objeto arbitrario compatible con Polars. Las series en Polars están diseñadas para ser eficientes y rápidas, aprovechando el rendimiento de Rust para manejar grandes volúmenes de datos de manera eficiente.

### Selección Condicional o Filtros

Una característica importante de polars es la selección condicional usando el método `.filter()`.

Los términos de búsqueda condicional o filtros se entregan al método como tipo `string`

````python
# Seleccionar todas las filas donde el valor que está en la columna "W" sea mayor que cero

df.filter(pl.col("W")>0) #
````

Para dos condiciones, se usa los booleanos de esta forma

* | en vez de or
* & en vez de and
* ~ en vez de not

````python
# Seleccionar las filas donde 'W' sea mayor que cero
# y también donde 'Y' sea mayor que 0.5

df.filter((pl.col('W') > 0) & (pl.col('Y') > 0.5))
````

````python
# Seleccionar las filas donde 'W' sea mayor que cero
# y de esas filas escoger los valores de la columna 'Y'

df.filter((pl.col('W') > 0))['Y']
````

````python
# Seleccionar las filas donde 'W' sea mayor que cero
# y de esas filas escoger los valores de las columna 'Y' y 'Z'

df.filter((pl.col('W') > 0))[['Y','Z']]
````

Para dos condiciones, puede usar `| = or` y `& = and` con paréntesis:

````python
df
````

````python
# Seleccionar las filas donde 'W' sea mayor que cero
# y también donde 'Y' sea mayor que 0.5

df.filter((pl.col('W') > 0) & (pl.col('Y') > 0.5))
````

### Agrupación por Filas

El método groupby le permite agrupar filas de datos y llamar a funciones agregadas.

````python
import polars as pl

# Crear dataframe desde un diccionario

data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB','GOOG','MSFT','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah','John','Randy','David'],
       'Sales':[200,120,340,124,243,350,275,400,180]}

data
````

````python
# Conversión del diccionario a dataframe

df = pl.DataFrame(data)
df
````

Se puede usar el método `.groupby()` para agrupar filas en función de un nombre de columna. Por ejemplo, vamos a agruparnos a partir de la Compañía. Esto creará un objeto `polars.dataframe.group_by.GroupBy`:

````python
#agrupar por Company

df.group_by('Company')
````

Se puede grabar este objeto en una nueva variable:

````python
by_comp = df.group_by("Company")
````

````python
by_comp
````

Utilizar los métodos agregados del objeto:

````python
# Promedio de todas las columnas por company
by_comp.mean()
````

````python
# Promedio de ventas por company, es bueno indicar las columnas numericas
# Para que tenga sentido
by_comp.mean()[["Company","Sales"]]
````

````python
# agrupar por compañia y calcular el promedio por cada una

df.group_by('Company').mean()[["Company","Sales"]]
````

Más ejemplos con métodos agregados:

````python
# agrupar por compañia y calcular la desviacion standard

by_comp.agg(pl.col('Sales').std())
````

````python
# agrupar por compañia y calcular el minimo

by_comp.min()[["Company","Sales"]]
````

````python
# agrupar por compañía y calcular el máximo

by_comp.max()[["Company","Sales"]]
````

````python
# agrupar por compañía y sumar los elementos que hay excluyendo los NaN

by_comp.len()
````

````python
# Para Generar estadísticas descriptivas que resuman la tendencia central,
# la dispersión y la forma de la distribución de un conjunto de datos,
# se pueden agrupar los calculos ccon el metodo .agg

by_comp.agg([
    pl.col("Sales").mean().alias("mean"),
    pl.col("Sales").std().alias("std"),
    pl.col("Sales").min().alias("min"),
    pl.col("Sales").max().alias("max"),
    pl.col("Sales").median().alias("median"),
    pl.col("Sales").count().alias("count")
])
````

## Pivot Tables

La funciónlidad “Pivot_table” es muy utilizada y popular en las conocidas “hojas de cálculo” tipo, OpenOffice, LibreOffice, Excel, Lotus, etc. Esta funcionalidad nos permite agrupar, ordenar, calcular datos y manejar datos de una forma muy similar a la que se hace con las hojas de cálculo. [Más información](https://docs.pola.rs/user-guide/transformations/pivot/).

La principal función del `.pivot` son las agrupaciones de datos a las que se les suelen aplicar funciones matemáticas como sumatorios, promedios, etc.

````python
import seaborn as sns # importar la librería seaborn
# cargar dataset del titanic
titanic = sns.load_dataset('titanic')
titanic.head()
````

````python
type(titanic)
````

````python
# convertir dataframe de pandas a polars

titanic = pl.from_pandas(titanic)
type(titanic)
````

````python
# Haciendo el Pivot table a mano para obtener el promedio de personas que sobrevivieron por genero

# 1. Agrupar por genero
# 2. Obtener los sobrevivientes
# 3. Calcular el promedio

titanic.group_by('sex').mean()[["sex", "survived"]]
````

````python
# Promedio de cuántos sobrevivieron por género divididos por clase

# 1. Agrupar por genero y clase
# 2. Calcular el promedio
# 3. Obtener los sobrevivientes

titanic.group_by(['sex', 'class']).mean()['sex', 'class','survived']
````

**Usando Pivot Tables**

````python
# Agrupar por 'sex' y 'class', calcular la media de 'survived' y hacer un pivot
grouped = titanic.group_by(['sex', 'class']).agg(
    pl.col('survived').mean()
)

# Crear una tabla dinámica (pivot table)
pivot_table = grouped.pivot(
    values='survived',
    index='sex',
    on='class'
)

pivot_table
````

### Concatenación

La concatenación básicamente combina DataFrames. Tenga en cuenta que las dimensiones deben coincidir a lo largo del eje con el que se está concatenando.

la Concatenación se hace con dataframes de diferentes indices

Puede usar `.concat()` y pasar una lista de DataFrames para concatenar juntos:

````python
# Concatenar cada dataframe verticalmente,
# ya que coinciden los nombres de las columnas

pl.concat([df1,df2,df3], how="vertical")
````

## Datos Categóricos

* Más información de datos categóricos:

### Creación

````python
# Creacion de una serie de datos categoricos
cate = pl.Series(["manzana", "banano", "corozo", "manzana","pera"],
                 dtype=pl.Categorical)
cate
````

````python
cate.dtype  # physical significa que el orden de las categorías está basado en el orden que se almacenan internamente
````

**Cuando trabajamos con columnas que tienen pocos valores únicos que se repiten, como tipos de fruta, podemos usar el tipo Categorical para ahorrar memoria y hacer los cálculos más rápidos. Es como si Polars en vez de guardar texto, guardara un número para cada fruta. Y ese ordering='physical' solo nos dice que Polars guarda las frutas en el orden en que aparecen, no alfabéticamente.**

````python
cate.describe()
````

````python
# Si quisieramos ordenarlos alfabeticamente, deberia de cambiarse el ordenamiento asi:

pl.Series(["manzana", "banano", "corozo", "manzana","pera"],
          dtype=pl.Categorical).cast(pl.Categorical(ordering="lexical"))
````

### Rellenar valores faltantes

````python
df
````

````python
# Llenar los datos faltantes con el dato que nos interese
df.fill_null("Llenar Valores")# llenar los espacios con un string
````

````python
# Llenar los datos faltantes con el dato que nos interese
df.fill_null(value=99) # llenar los espacios con un numero
````

🧠 **¿Por qué la columna D no se llena con 99?**

💥 **Porque la columna D es de tipo NullType**

Polars no sabe qué tipo de dato debería tener D, porque todos sus valores son null, y no tiene forma de inferir que debería ser un número.

````python
print(df.schema)
````

✅ **¿Cómo solucionarlo?**

Debes convertir explícitamente la columna D a un tipo válido antes de usar fill_null. Por ejemplo:

````python
df = df.with_columns(
    pl.col("D").cast(pl.Int64)
)

df_filled = df.fill_null(99)
print(df_filled)
````

#### Rellenar valores faltantes con estadístico

````python
import polars as pl

df = pl.DataFrame({
    "A": [1, 2, None],
    "B": [5, None, None],
    "C": [1, 2, 3],
})
````

````python
print(df)
````

🧮 **Opción 1: Llenar con el promedio (mean)**

````python
df_filled_mean = df.with_columns([
    pl.col("A").fill_null(pl.col("A").mean()),
    pl.col("B").fill_null(pl.col("B").mean()),
])

print(df_filled_mean)
````

📈 **Opción 2: Llenar con la mediana (median)**

````python
df_filled_median = df.with_columns([
    pl.col("A").fill_null(pl.col("A").median()),
    pl.col("B").fill_null(pl.col("B").median()),
])

df_filled_median
````

✅ **¿Qué pasa con columnas que no tienen nulls?**

No pasa nada. .fill_null(...) simplemente no afecta esas columnas.

````python
# Separar columnas por tipo

numeric_cols = [col for col, dtype in df.schema.items()
                if dtype in (pl.Int64, pl.Float64, pl.Int32, pl.Float32)]

non_numeric_cols = [col for col in df.columns if col not in numeric_cols]
````

**Explicación del `list_comprehension`**:

Es un list comprehension en Python, que:

```
col for col, dtype in df.schema.items()
                if dtype in (pl.Int64, pl.Float64, pl.Int32, pl.Float32)
```

* Recorre todas las columnas del DataFrame con sus tipos (col, dtype).

* Filtra solo las que son numéricas (Int o Float).

* Devuelve una lista de nombres de columnas numéricas.


**Estamos sacando solo los nombres de las columnas que tienen tipo numérico, para poder aplicar una estrategia diferente de llenado (como el promedio). El `for col, dtype in df.schema.items()` nos da los nombres y tipos de todas las columnas, y el `if dtype in (...)` se asegura de que tomemos solo las que son numéricas.**

````python
# Aplicar estrategias de llenado

df_filled = df.with_columns([

    # Llenar numéricas con su promedio
    *[pl.col(col).fill_null(pl.col(col).mean()) for col in numeric_cols],

    # Llenar no numéricas con un string fijo
    *[pl.col(col).fill_null("desconocido") for col in non_numeric_cols]
])

print(df_filled.head())
````

**¿Qué significa el asterisco (*)?**

```
*[pl.col(col).fill_null(pl.col(col).mean()) for col in numeric_cols]
```


🔍 **¿Qué hace el * aquí?**

Este asterisco es el **operador de "unpacking" de listas.**

❗ Sin el *, estarías pasando una lista de expresiones. ✅ Con el *, estás pasando varias expresiones separadas, una por una.



🧠 **Explicación simple:**
Supón que haces esto:

```
acciones = [
    pl.col("edad").fill_null(pl.col("edad").mean()),
    pl.col("ingresos").fill_null(pl.col("ingresos").mean())
]
```

Y luego:

```
df.with_columns(acciones)  # Esto funciona
```

Pero si tienes que crear esa lista automáticamente:

```
acciones = [pl.col(col).fill_null(pl.col(col).mean()) for col in numeric_cols]
```

Entonces para pasar cada expresión individualmente a `.with_columns()`, haces:

`df.with_columns(*acciones)`

Y si quieres hacerlo todo en una sola línea, usas directamente:

```
df.with_columns(
    *[pl.col(col).fill_null(pl.col(col).mean()) for col in numeric_cols]
)
```

🧠 **El * en Python se usa para 'desempacar' una lista. En este caso, la función .`with_columns()` espera recibir muchas expresiones, no una lista con todas ellas. Al poner * delante del bucle, le estamos diciendo a Python: _Toma cada elemento de esta lista y pásalo por separado_. Así evitamos errores y el código queda más limpio.**

# Algunas Conclusiones

* 🧾 **¿Qué es un schema en Polars?**

    * Un schema es simplemente una descripción de la estructura de un DataFrame: muestra los nombres de las columnas y sus tipos de datos.

    * 🔎 **Puedes pensarlo como:**

        Un "resumen técnico" del contenido del DataFrame:
        ❝¿Qué columnas tengo y qué tipo de datos hay en cada una?❞

* **En Polars, las operaciones no son "inplace" por defecto, porque los DataFrame son inmutables.**
    * Esto significa que cada operación devuelve una nueva copia modificada, y no cambia el objeto original.

    * 📌 **¿Por qué no hay inplace=True como en Pandas?**
        Porque Polars está diseñado para ser:

        * más eficiente (al evitar modificar datos en memoria),

        * más predecible (sin efectos secundarios),

        * y más amigable para trabajar en paralelo o con grandes volúmenes de datos.

### 📊 Comparativa entre **Polars** y **Pandas**

| Característica                  | Pandas                                      | Polars                                                    |
|--------------------------------|----------------------------------------------|------------------------------------------------------------|
| **Mutabilidad**                | DataFrames mutables (se modifican directamente) | DataFrames inmutables (hay que reasignar el resultado)    |
| **Métodos sobre columnas**     | `df["col"].mean()`                           | `pl.col("col").mean()` dentro de `.select()` o `.with_columns()` |
| **Método `.info()`**           | ✅ Disponible (`df.info()`)                   | ❌ No existe `.info()`, se usa `.schema` o `.describe()`   |
| **Tipo de operaciones**        | Imperativas (paso a paso)                    | Declarativas y expresivas (basadas en expresiones)         |
| **Rendimiento**                | Bueno, pero limitado en datos muy grandes    | 🚀 Muy alto, optimizado para grandes volúmenes de datos     |
