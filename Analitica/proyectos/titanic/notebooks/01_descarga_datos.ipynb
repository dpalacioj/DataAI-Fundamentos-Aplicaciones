# 1. Descarga de los datos

El primer paso consiste en descargar el conjunto de datos y seleccionar las columnas que serán usadas en el proyecto

## Importar librerías

En la página [OpenML]('https://openml.org/search?type=data&status=active') podrás encontrar diferentes conjuntos de datos *Open Source* para usarlos en proyectos personales o ensayar distintos enfoques de *machine learning*.

```python
# Librerías base

from pathlib import Path
import pandas as pd
```

## Carga de datos

```python
url_datos = "https://www.openml.org/data/get_csv/16826755/phpMYEkMl"

datos_titanic = pd.read_csv(url_datos, low_memory=False)

datos_titanic.info()
```

Para ver en detalle 10 filas de manera "aleatoria" basta con ejecutar el siguiente script:

```python
datos_titanic.sample(10)
```

## Información del conjunto de datos

La finalidad de este proyecto es predecir si un pasajero del **Titanic** sobrevirá o no.

Para esto, es fundamental entender qué tipo de variables se tienen:

```python
datos_titanic.columns
```

Las siguientes variables se relacionan con la variable "survived":

* `boat`: El número del bote salvavidas (si sobrevivieron)
* `body`: El número del cuerpo (si no sobrevivieron y el cuerpo fue recuperado)

**Tener presente** un concepto fundamental que se conoce como ***data leakage***, el cual significa que algunas columnas podrían traer consigo la variable que vamos a predecir. Por ejemplo: si conocemos el número del bote salvavidas, esto implica que el pasajero sobrevivió, por lo tanto, la variable "survived" ya se conocería de manera implícita.

Por lo tanto, las columnas `boat` y `body` no se usarán para evitar introducir *data leakage* en la información:

```python
columnas_a_usar = [
    "pclass",
    "survived",
    "name",
    "sex",
    "age",
    "sibsp",
    "parch",
    "ticket",
    "fare",
    "cabin",
    "embarked",
    "home.dest",
]
```

## Datos finales

```python
# Para tenerlos en formato csv
datos_titanic_final = pd.read_csv(url_datos, usecols=columnas_a_usar, low_memory=False)

# Verificamos que se hayan cargado las columnas que queríamos
datos_titanic_final.info()
```

Como era de esperarse, se removió `boat` y `body`.

## Descargar y guardar los datos en formato `parquet`

```python
# Obtener la ruta del notebook actual
ruta_datos_crudos = Path.cwd().resolve().parent / "datos"

ruta_archivo = ruta_datos_crudos / "01_datos_crudos_titanic.parquet"

# Crear directorio si no existe
ruta_datos_crudos.mkdir(parents=True, exist_ok=True)

# Guardar el DataFrame en el archivo parquet
datos_titanic_final.to_parquet(ruta_archivo, index=False)
```

Finalmente el conjunto de datos resultante fue guardado en formato parquet y está listo para la exploración de los datos.
