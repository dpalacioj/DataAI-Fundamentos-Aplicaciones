# ¿Qué es ngrok y qué hace?

**ngrok** es una herramienta que crea un túnel seguro desde una máquina local a Internet. Esto permite exponer de forma temporal y segura servicios o aplicaciones que se ejecutan localmente, como una aplicación de Streamlit.

Entre sus usos destacan:

- **Desarrollo y pruebas:** Permite compartir y probar aplicaciones en desarrollo sin necesidad de desplegarlas en un servidor público.
- **Demostraciones remotas:** Facilita la demostración de aplicaciones a usuarios externos o clientes de forma rápida.
- **Acceso en entornos restringidos:** Es útil en entornos como Google Colab para exponer un servidor local a través de una URL pública.

# Seguridad del Authtoken de ngrok

El **authtoken** es una clave privada que vincula el uso de ngrok con tu cuenta personal. Con este token, ngrok puede:

- **Autenticar y registrar** tu uso del servicio.
- **Aplicar configuraciones personalizadas** y mantener la seguridad de los túneles.

## ¿Es inseguro exponer el authtoken?

Sí, exponer tu authtoken públicamente puede ser riesgoso porque:

- **Acceso no autorizado:** Podrían utilizar tu token para acceder a tu cuenta de ngrok y abusar de tus túneles.
- **Riesgo de abuso:** Terceros podrían generar tráfico no deseado o incurrir en costos si se superan los límites del servicio.

**Recomendación:**  
Mantén tu authtoken privado y evita compartirlo en repositorios públicos o foros accesibles. Utiliza variables de entorno u otros métodos seguros para gestionar tus credenciales.

```python
from pyngrok import ngrok
import os
from dotenv import load_dotenv
```

```python
env_path = '/content/drive/MyDrive/cursos-para-dictar/UDM/.env'

# Carga las variables de entorno definidas en el archivo
load_dotenv(env_path)

# Configura ngrok con el token obtenido del archivo .env
ngrok.set_auth_token(os.environ['NGROK_AUTHTOKEN'])

# Iniciar un túnel en el puerto 8501
public_url = ngrok.connect(8501)

print("Tu aplicación Streamlit está disponible en:", public_url)
```
